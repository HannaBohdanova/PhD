```python
import numpy as np

# Визначаємо матрицю A та вектор x
A = np.array([
    [1/3, 1/3, 1/3],
    [0, 3/4, 1/4],
    [1/2, 0, 1/2]
])

x = np.array([1, 0, 0])

iteration = 0
previous_x = None
#np.set_printoptions(precision=4, suppress=True)
while True:
    iteration += 1
    x_new = A @ x
    print(f"Iteration {iteration}: {x_new}")

    if previous_x is not None and np.linalg.norm(x_new - previous_x) < 1e-10:
        print("Зупинка: отримано два однакових результати поспіль.")
        break

    previous_x = x_new.copy()
    x = x_new
    
print(x_new, previous_x)
```

```python
import numpy as np

# Визначаємо матрицю A та вектор x
A = np.array([
    [1, 0, 0],
    [1/3, 1/3, 1/3],
    [1/5, 0, 4/5]
])
x = np.array([1, 0, 1/2])
result = A @ x   # або np.dot(W, x)

print(result)
# Параметри зупинки
epsilon = 1e-6   # точність
max_iter = 1000  # захист від нескінченного циклу

for t in range(max_iter):
    x_new = A @ x

    # перевірка консенсусу: всі значення майже однакові
    if np.max(x_new) - np.min(x_new) < epsilon:
        print(f"Консенсус досягнуто на ітерації {t+1}")
        break

    x = x_new

print("Кінцевий вектор думок:", x_new)
```

```python
import numpy as np

# Приклад стохастичної матриці
A = np.array([
    [1, 0, 0],
    [1/3, 1/3, 1/3],
    [1/2, 0, 1/2]
])
np.set_printoptions(precision=4, suppress=True)
# Починаємо з одиничної
A_power = np.identity(A.shape[0])

prev = None
max_iter = 1000
tol = 1e-10  # точність
for i in range(1, max_iter + 1):
    A_power = A_power @ A
    if prev is not None and np.allclose(A_power, prev, atol=tol):
        print(f"Досягнута збіжність на кроці {i}")
        break
    prev = A_power.copy()

print("T^∞ ≈")
print(A_power)
```
