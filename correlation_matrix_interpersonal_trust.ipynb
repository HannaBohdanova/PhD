```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy.cluster.hierarchy import dendrogram, linkage, fcluster
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.DataFrame(result)

# Масштабування (бо показники в різних одиницях)
df = df.drop(columns=['Year','ISO3', 'Country_y', 'Country'])
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
print (X_scaled)
```

```python
# Метод linkage – ward, average або complete
Z = linkage(X_scaled, method="ward")
```

```python
plt.figure(figsize=(8, 5))
dendrogram(Z, leaf_rotation=45)
plt.title("Ієрархічна кластеризація (дендрограма)")
plt.xlabel("Країни")
plt.ylabel("Відстань")
plt.savefig("dendrogram.png")
```

```python
# Наприклад, хочемо 2 кластери
clusters = fcluster(Z, t=2, criterion="maxclust")
df["Cluster"] = clusters
print(df)
```

```python
plt.close()
plt.clf()
sns.scatterplot(
    x="Interpersonal_trust", y="CPI",
    hue="Cluster", data=df, palette="deep", s=100
)
plt.title("Кластери країн за довірою та якістю життя")
plt.savefig("Cluster2.png")
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
plt.close()
plt.clf()
# Список для збереження "inertia" для різних k
inertia = []

# Перевіримо від 1 до 10 кластерів
K = range(1, 11)

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(df)
    inertia.append(kmeans.inertia_)

# Побудова графіка
plt.plot(K, inertia, 'bo-')
plt.xlabel('The Elbow Method')
plt.ylabel('WCSS')
plt.title('Number of clusters')
plt.savefig("Метод ліктя для вибору k.png")
```

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# Припустимо, що ваші числові дані вже у DataFrame "df"
# (без назв країн у стовпцях, тільки числа)
plt.close()
plt.clf()
print(df)
silhouette_scores = []
K = range(2, 11)  # перевіряємо від 2 до 10 кластерів

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(df)
    score = silhouette_score(df, labels)
    silhouette_scores.append(score)

# Побудова графіка
plt.plot(K, silhouette_scores, 'bo-')
plt.xlabel('Number of clusters')
plt.ylabel('Silhouette coefficients')
plt.title('Silhouette Method')
plt.savefig("Метод силуета для вибору k.png")
```
